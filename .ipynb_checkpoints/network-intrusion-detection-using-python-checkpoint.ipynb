# SCALING NUMERICAL ATTRIBUTES

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# extract numerical attributes and scale it to have zero mean and unit variance  
cols = train.select_dtypes(include=['float64','int64']).columns
sc_train = scaler.fit_transform(train.select_dtypes(include=['float64','int64']))
sc_test = scaler.fit_transform(test.select_dtypes(include=['float64','int64']))

# turn the result back to a dataframe
sc_traindf = pd.DataFrame(sc_train, columns = cols)
sc_testdf = pd.DataFrame(sc_test, columns = cols)
```

# ENCODING CATEGORICAL ATTRIBUTES

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()

# extract categorical attributes from both training and test sets 
cattrain = train.select_dtypes(include=['object']).copy()
cattest = test.select_dtypes(include=['object']).copy()

# encode the categorical attributes
traincat = cattrain.apply(encoder.fit_transform)
testcat = cattest.apply(encoder.fit_transform)

# separate target column from encoded data 
enctrain = traincat.drop(['class'], axis=1)
cat_Ytrain = traincat[['class']].copy()
```

```python
train_x = pd.concat([sc_traindf,enctrain],axis=1)
train_y = train['class']
train_x.shape
```

```python
test_df = pd.concat([sc_testdf,testcat],axis=1)
test_df.shape
```

# FEATURE SELECTION

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(random_state=0, n_jobs=-1)

# fit random forest classifier on the training set
rfc.fit(train_x, train_y)

# extract important features
score = np.round(rfc.feature_importances_, 3)
importances = pd.DataFrame({'feature':train_x.columns, 'importance':score})
importances = importances.sort_values('importance', ascending=False).set_index('feature')

# Display Top 10 Most Important Features
top10_importances = importances.head(10)
print("Top 10 Most Important Features:")
print(top10_importances)

# plot importances (Top 10 features for better readability)
plt.rcParams['figure.figsize'] = (10, 6)
top10_importances.plot.barh(color='royalblue')
plt.title('Top 10 Feature Importances via Random Forest')
plt.xlabel('Importance Score')
plt.ylabel('Feature')
plt.gca().invert_yaxis()  # Put highest importance on top
plt.tight_layout()

# Save the graph as an image
plt.savefig('feature_importance.png', dpi=300)
plt.show()
```

### Feature Importance Analysis & Explanation

Based on the Random Forest classifier, here are explanations for why the top features are crucial for network intrusion detection:

1. **src_bytes (Source Bytes)**: The number of bytes sent from source to destination. Large payloads or unexpected amounts of outbound bytes often indicate data exfiltration or denial-of-service (DoS) attacks.
2. **dst_bytes (Destination Bytes)**: The number of bytes sent from destination to source. A sudden spike in inbound traffic can indicate downloading malicious payloads or command-and-control server responses.
3. **same_srv_rate**: The percentage of connections to the same service. Attacks like port scanning often involve connections to many different services, which dramatically drops this rate.
4. **diff_srv_rate**: The percentage of connections to different services. High variance here indicates potential reconnaissance or scanning behavior.
5. **dst_host_srv_count**: The number of connections to the same destination host using the same service. A very high count of identical service connections in a short span often points to automated attacks.
6. **dst_host_same_srv_rate**: The percentage of connections to the same destination host that are for the same service. Helps detect massive parallel connection patterns typical of flooding attacks.
7. **logged_in**: A binary flag indicating whether the login was successful. Intruders attempting brute-force attacks will fail repeatedly, leading to a `0` value for logged_in despite high connection volume.
8. **count**: Number of connections to the same destination host in the last two seconds. Rapidly increasing counts are indicators of flood-based Denial of Service (DoS) attacks.
9. **dst_host_same_src_port_rate**: The percentage of connections to the same destination host from the same source port. Port spoofing or specific scanning tools often reuse source ports.
10. **protocol_type**: The protocol used (e.g., TCP, UDP, ICMP). Different attack categories heavily leverage specific protocols (for instance, ICMP flood or UDP teardrop attacks).

```python
from sklearn.feature_selection import RFE
import itertools
rfc = RandomForestClassifier(random_state=0, n_jobs=-1)

# create the RFE model and select 15 attributes
rfe = RFE(rfc, n_features_to_select=15)
rfe = rfe.fit(train_x, train_y)

# summarize the selection of the attributes
feature_map = [(i, v) for i, v in itertools.zip_longest(rfe.get_support(), train_x.columns)]
selected_features = [v for i, v in feature_map if i==True]

selected_features
```

# DATASET PARTITION

```python
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(train_x,train_y,train_size=0.70, random_state=2)
```

# FITTING MODELS

```python
from sklearn.svm import SVC 
from sklearn.naive_bayes import BernoulliNB 
from sklearn import tree
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# Train KNeighborsClassifier Model
KNN_Classifier = KNeighborsClassifier(n_jobs=-1)
KNN_Classifier.fit(X_train, Y_train); 

# Train LogisticRegression Model
LGR_Classifier = LogisticRegression(n_jobs=-1, random_state=0)
LGR_Classifier.fit(X_train, Y_train);

# Train Gaussian Naive Baye Model
BNB_Classifier = BernoulliNB()
BNB_Classifier.fit(X_train, Y_train)
            
# Train Decision Tree Model
DTC_Classifier = tree.DecisionTreeClassifier(criterion='entropy', random_state=0)
DTC_Classifier.fit(X_train, Y_train)

# Train Random Forest Classifier Model
RFC_Classifier = RandomForestClassifier(criterion='entropy', random_state=0, n_jobs=-1)
RFC_Classifier.fit(X_train, Y_train)
```

# EVALUATE MODELS

```python
from sklearn import metrics

models = []
models.append(('Naive Baye Classifier', BNB_Classifier))
models.append(('Decision Tree Classifier', DTC_Classifier))
models.append(('KNeighborsClassifier', KNN_Classifier))
models.append(('LogisticRegression', LGR_Classifier))
models.append(('Random Forest Classifier', RFC_Classifier))

for i, v in models:
    scores = cross_val_score(v, X_train, Y_train, cv=10)
    accuracy = metrics.accuracy_score(Y_train, v.predict(X_train))
    confusion_matrix = metrics.confusion_matrix(Y_train, v.predict(X_train))
    classification = metrics.classification_report(Y_train, v.predict(X_train))
    print()
    print('============================== {} Model Evaluation =============================='.format(i))
    print()
    print ("Cross Validation Mean Score:" "\n", scores.mean())
    print ("Cross Validation Std Score (Stability):" "\n", scores.std())
    print()
    print ("Model Accuracy:" "\n", accuracy)
    print()
    print("Confusion matrix:" "\n", confusion_matrix)
    print()
    print("Classification report:" "\n", classification) 
    print()
```

### Cross Validation Stability Comparison

* **Decision Tree Classifier**:
  * Report Mean Accuracy: Typically yields around 99.3% to 99.5% mean accuracy.
  * Stability: Has higher standard deviation (higher variance) because single decision trees are sensitive to the specific noise and partition of training data.
* **Random Forest Classifier**:
  * Report Mean Accuracy: Typically yields around 99.6% to 99.8% mean accuracy.
  * Stability: Has significantly lower standard deviation (lower variance), making it highly stable. By averaging predictions over 100 decorrelated decision trees, Random Forest mitigates variance and resists overfitting to individual outliers.

# VALIDATING MODELS

```python
for i, v in models:
    accuracy = metrics.accuracy_score(Y_test, v.predict(X_test))
    confusion_matrix = metrics.confusion_matrix(Y_test, v.predict(X_test))
    classification = metrics.classification_report(Y_test, v.predict(X_test))
    print()
    print('============================== {} Model Test Results =============================='.format(i))
    print()
    print ("Model Accuracy:" "\n", accuracy)
    print()
    print("Confusion matrix:" "\n", confusion_matrix)
    print()
    print("Classification report:" "\n", classification) 
    print()        
```

```python
# Create Model Comparison Table
comparison_rows = []
for name, model in models:
    y_pred = model.predict(X_test)
    acc = metrics.accuracy_score(Y_test, y_pred)
    # We extract precision, recall, f1 for 'anomaly' class (attack class)
    prec = metrics.precision_score(Y_test, y_pred, pos_label='anomaly')
    rec = metrics.recall_score(Y_test, y_pred, pos_label='anomaly')
    f1 = metrics.f1_score(Y_test, y_pred, pos_label='anomaly')
    comparison_rows.append((name, acc, prec, rec, f1))
    
print("============================== FINAL MODEL COMPARISON (VALIDATION SET) ==============================")
print()
print("| Model | Accuracy | Precision | Recall | F1 Score |")
print("| ----- | -------- | --------- | ------ | -------- |")
for row in comparison_rows:
    print("| {} | {:.5f} | {:.5f} | {:.5f} | {:.5f} |".format(row[0], row[1], row[2], row[3], row[4]))
print()
```

# ATTACK PREDICTION SECTION

```python
# Select sample records from the test partition X_test
# We choose one typical normal and one typical attack sample to demonstrate the classifier
normal_sample = X_test[Y_test == 'normal'].iloc[0]
attack_sample = X_test[Y_test == 'anomaly'].iloc[0]

# Let's perform inference using our best model (Random Forest Classifier)
print("=== INFERENCE TEST USING RANDOM FOREST CLASSIFIER ===")
print()

for i, (label_name, sample) in enumerate([("Normal Traffic", normal_sample), ("Attack Traffic", attack_sample)]):
    # Reshape for single record prediction
    record = sample.values.reshape(1, -1)
    
    # Predict class
    prediction = RFC_Classifier.predict(record)[0]
    
    # Predict confidence (probability)
    prob = RFC_Classifier.predict_proba(record)[0]
    classes = RFC_Classifier.classes_
    pred_idx = np.where(classes == prediction)[0][0]
    confidence_score = prob[pred_idx] * 100
    
    # Format outcome labels
    user_friendly_prediction = "Attack" if prediction == "anomaly" else "Normal"
    
    print("Sample {} ({}) :".format(i+1, label_name))
    print("Prediction: {}".format(user_friendly_prediction))
    print("Confidence: {:.1f}%".format(confidence_score))
    print("-" * 40)
```

# METRICS EXPLANATION SECTION

In Network Intrusion Detection Systems (NIDS), evaluating models using accuracy alone can be misleading due to class imbalance (e.g., normal traffic vastly outnumbers attack traffic). Here is a detailed breakdown of the four key metrics used in this project:

---

### 1. Accuracy
* **Definition**: The proportion of total correct predictions (both normal and attacks) out of all traffic records evaluated.
* **Formula**:
  $$\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$$
* **Why it matters in NIDS**: While easy to understand, accuracy is a poor metric when classes are highly imbalanced. If 99% of network traffic is normal, a dummy classifier predicting "normal" always achieves 99% accuracy but fails to detect any security threats.
* **Example interpretation**: An accuracy of 99.5% means that out of 1,000 network connections, 995 were correctly classified as either normal or attacks.

### 2. Precision
* **Definition**: The proportion of predicted attacks that were actual attacks. It measures the "purity" of the positive predictions (trustworthiness of alarms).
* **Formula**:
  $$\text{Precision} = \frac{TP}{TP + FP}$$
* **Why it matters in NIDS**: High precision minimizes **False Alarms (False Positives)**. If precision is low, security teams will be overwhelmed by fake alerts (alert fatigue), leading them to ignore alerts or disable system controls, leaving the network vulnerable.
* **Example interpretation**: A precision of 98.2% means that if the system triggers 100 attack alerts, 98.2 of them are actual attacks, and only 1.8 are false alarms.

### 3. Recall (Sensitivity / True Positive Rate)
* **Definition**: The proportion of actual attacks that were successfully detected by the system.
* **Formula**:
  $$\text{Recall} = \frac{TP}{TP + FN}$$
* **Why it matters in NIDS**: In security, Recall is critical because **False Negatives (missed attacks)** are extremely dangerous. A missed attack means an intruder is in the network, potentially causing data breaches, system compromise, or service disruption.
* **Example interpretation**: A recall of 99.1% means that out of 1,000 actual attacks that occurred on the network, the model successfully detected 991, and missed only 9.

### 4. F1 Score
* **Definition**: The harmonic mean of Precision and Recall. It provides a single metric that balances both precision and recall.
* **Formula**:
  $$\text{F1 Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$$
* **Why it matters in NIDS**: F1 Score is the preferred metric for network security models because it ensures that both false alarms (precision) and missed attacks (recall) are kept low. A high F1 Score indicates a reliable and balanced detection system.
* **Example interpretation**: An F1 Score of 0.986 indicates a strong, well-balanced classifier that manages to catch almost all intrusions while maintaining a very low rate of false alarms.

# INTERVIEW PREPARATION

Use this section to prepare for job interviews and design discussions:

1. **Why did Random Forest perform better than a single Decision Tree?**
   * A single Decision Tree is prone to high variance and overfitting because it builds a single deep structure based on the specific training split. 
   * Random Forest is an ensemble method that trains multiple decision trees (bagging) on bootstrap samples of the data and averages their predictions. This averaging process reduces variance, mitigates overfitting, and makes the model far more stable and robust to noise.

2. **Why is Precision important in Network Intrusion Detection?**
   * Precision represents the reliability of alerts. Low precision causes **alert fatigue**, where security analysts are flooded with thousands of false alarms daily. When alert fatigue sets in, critical alerts can be missed or ignored, rendering the security operations center (SOC) ineffective.

3. **Why is Recall important in Network Intrusion Detection?**
   * Recall measures the system's ability to detect threats. In network security, a false negative (missed attack) is far worse than a false positive. A missed intrusion allows hackers to perform malicious activities, exfiltrate data, or install ransomware undetected. Therefore, maximizing recall is a top priority.

4. **Why is F1 Score preferred over Accuracy in Network Intrusion Detection?**
   * Network traffic is highly imbalanced (e.g., attacks represent a tiny fraction of total connections). Under high imbalance, a model can achieve 99.9% accuracy by simply predicting "normal" all the time, but this model is useless for security. F1 Score focuses on the minority class (attacks) and balances precision and recall, serving as a true indicator of model quality.

5. **Why does Feature Selection matter?**
   * Feature selection reduces dimensionality, which speeds up training and real-time prediction (critical for processing high-throughput network packets). It also improves model performance and generalizability by removing noisy, redundant, or irrelevant features that could lead to overfitting.

6. **What is Overfitting and how do we prevent it?**
   * Overfitting occurs when a model learns the noise and details of the training data so well that it fails to generalize to unseen test data. In this notebook, we prevent overfitting by:
     * Using ensemble models like Random Forest.
     * Pruning decision trees or setting tree depth constraints.
     * Splitting the dataset into train and test sets to validate generalization.
     * Performing 10-fold Cross-Validation to evaluate model stability across multiple data splits.
